```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier,
    AdaBoostClassifier,
    GradientBoostingClassifier
)
from sklearn.discriminant_analysis import (
    LinearDiscriminantAnalysis,
    QuadraticDiscriminantAnalysis
)
from sklearn.naive_bayes import GaussianNB
from sklearn.neural_network import MLPClassifier
```

```python
RUTA_CSV = r"C:\Users\manue\Downloads\3ccbe\Proyecto_final\Aves\5avesC_limpio.csv"
SALIDA_RESULTADOS = r"C:\Users\manue\Downloads\3ccbe\Proyecto_final\Aves\5Resultado_LB1.csv"

df = pd.read_csv(RUTA_CSV)
print(f"Registros : {len(df)}")
print(f"Columnas  : {len(df.columns)}")

# Todas las columnas numéricas excepto label
X = df.select_dtypes(include="number").drop(columns=["label"])
# Variable objetivo
y = df["label"]
print(f"Número de características : {X.shape[1]}")
print(f"Número de clases          : {y.nunique()}")
```

Verificacion

```python
conteo = y.value_counts()
print(conteo[conteo < 2])
```

Train Test y Modelo

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,
    test_size=0.30,random_state=1024,stratify=y)
print("\nTrain:", len(X_train))
print("Test :", len(X_test))


modelos = {

    "Logistic Regression":
        LogisticRegression(max_iter=1000),

    "KNN":
        KNeighborsClassifier(),

    "SVM":
        SVC(),

    "Decision Tree":
        DecisionTreeClassifier(random_state=1024),

    "Random Forest":
        RandomForestClassifier(random_state=1024),

    "Extra Trees":
        ExtraTreesClassifier(random_state=1024),

    "AdaBoost":
        AdaBoostClassifier(random_state=1024),

    "Gradient Boosting":
        GradientBoostingClassifier(random_state=1024),

    "LDA":
        LinearDiscriminantAnalysis(),
        
    "Naive Bayes":
        GaussianNB(),

    "ANN":
        MLPClassifier(
            hidden_layer_sizes=(100,),
            max_iter=500,
            random_state=1024
        )
}
```

Entrenamineto

```python
resultados = []

print("\n"+"="*60)
print("ENTRENANDO MODELOS")
print("="*60)

for nombre, modelo in modelos.items():

    print(f"\nEntrenando {nombre}...")

    modelo.fit(X_train, y_train)

    pred = modelo.predict(X_test)

    accuracy = accuracy_score(y_test, pred)

    precision = precision_score(
        y_test,
        pred,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        y_test,
        pred,
        average="weighted",
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        pred,
        average="weighted",
        zero_division=0
    )

    print(f"Accuracy : {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall   : {recall:.4f}")
    print(f"F1 Score : {f1:.4f}")

    resultados.append({

        "Modelo": nombre,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1

    })
```

Tabla de resultados

```python
resultados = pd.DataFrame(resultados)

resultados = resultados.sort_values(
    by="Accuracy",
    ascending=False
)

resultados.to_csv(
    SALIDA_RESULTADOS,
    index=False
)

print("\n"+"="*60)
print("RESULTADOS FINALES")
print("="*60)

print(resultados)

print("\nArchivo generado:")
print(SALIDA_RESULTADOS)

print("\nPROCESO TERMINADO")
```
